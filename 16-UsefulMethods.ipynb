enumerate:

```python
index = 0
for number in list(range(5,10)):
    print(f"no: {number} ix: {index}")
    index += 1
```

```python
for element  in enumerate(list(range(5,15))):
    print(element)
    
```

```python
for (index,number) in enumerate(list(range(5,10))):
    print(number)
    
```
