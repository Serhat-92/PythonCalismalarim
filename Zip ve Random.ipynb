## zip:

```python
sport_list = ["run", "swim", "basketball"]
```

```python
calorie_list = [100, 200, 300]
```

```python
day_list = ["monday:", "tuesday:", "wendesday:"]
```

```python
new_list = list(zip(sport_list,calorie_list,day_list))
```

```python
new_list
```

```python
for (x,y,z) in new_list:
    print(z,x,y)
```

## list advanced

```python
my_list_2 = []
my_string = "metallica"

for element in my_string:
    my_list_2.append(element)
```

```python
my_list_2
```
